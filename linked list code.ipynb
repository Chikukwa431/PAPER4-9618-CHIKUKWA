```python
class Node:
    def __init__(self, data):
        self.__data = data
        self.__pointer = None

    def setData(self, data):
        self.__data = data

    def setPointer(self, pointer):
        self.__pointer = pointer

    def getData(self):
        return self.__data

    def getPointer(self):
        return self.__pointer


class LinkedList:
    def __init__(self):
        self.__StartNode = None

    def InsertNode(self, new_node: Node):
        # Insert at the beginning of the list
        new_node.setPointer(self.__StartNode)
        self.__StartNode = new_node

    def SearchNode(self, search_data):
        current = self.__StartNode
        found = False
        
        while current is not None:
            if current.getData() == search_data:
                found = True
                break
            current = current.getPointer()
            
        if found:
            print(f"Data '{search_data}' was found in the linked list.")
        else:
            print(f"Data '{search_data}' was not found in the linked list.")

    def printNode(self):
        current = self.__StartNode
        if current is None:
            print("The linked list is empty.")
            return
        
        print("Linked List Contents:")
        while current is not None:
            print(current.getData(), end=" -> ")
            current = current.getPointer()
        print("None")


# --- Example Usage / Testing ---
if __name__ == "__main__":
    # Create the linked list
    my_list = LinkedList()

    # Create individual nodes
    nd1 = Node("dog")
    nd2 = Node("cat")
    nd3 = Node("mouse")

    # Insert nodes into the list
    my_list.InsertNode(nd1)
    my_list.InsertNode(nd2)
    my_list.InsertNode(nd3)

    # Print all nodes in the list
    my_list.printNode()

    # Search for nodes
    my_list.SearchNode("dog")
    my_list.SearchNode("elephant")
```
