```python
# Question1_N23

def IterativeVowels(Value):
    """
    Function to count lower-case vowels in a string using an iterative approach.
    Based on the pseudocode provided in the task.
    """
    Total = 0
    LengthString = len(Value)
    
    # Iterate through the string based on its length 
    for x in range(0, LengthString):
        # Get the first character of the current string
        FirstCharacter = Value[0]
        
        # Check if the character is a lower-case vowel 
        if FirstCharacter == 'a' or FirstCharacter == 'e' or \
           FirstCharacter == 'i' or FirstCharacter == 'o' or \
           FirstCharacter == 'u':
            Total = Total + 1
            
        # Update the string by removing the first character 
        Value = Value[1:]
        
    return Total

def RecursiveVowels(Value):
    """
    Function to count lower-case vowels in a string using a recursive approach.
    """
    # Base case: if the string is empty, return 0
    if len(Value) == 0:
        return 0
    else:
        FirstCharacter = Value[0]
        # Check if the first character is a vowel
        if FirstCharacter in ['a', 'e', 'i', 'o', 'u']:
            return 1 + RecursiveVowels(Value[1:])
        else:
           return 0 + RecursiveVowels(Value[1:])

# --- Main Program Calls ---

# Task 1(a)(ii): Call IterativeVowels with "house" and output result [cite: 74, 75]
print("Iterative result for 'house':", IterativeVowels("house"))

# Task 1(b)(ii): Call RecursiveVowels with "imagine" and output result [cite: 89, 90]
print("Recursive result for 'imagine':", RecursiveVowels("imagine"))
```
