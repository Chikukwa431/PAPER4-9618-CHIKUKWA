```python

# Base Case (len(s) <= 1): Recursion must stop when it reaches the smallest possible sub-problem. 
# If a string has 0 or 1 character, it is already reversed, so the function returns it immediately.
# Extraction (s[-1]): This retrieves the final character of the current string.
# Reduction (s[:-1]): This slices the string to exclude the last character, reducing the problem size with every recursive iteration.
# Combination (+): The extracted last character is concatenated to the front of the result returned by the next recursive call, 
# systematically building the reversed string from back to front.

def ReverseString(s):
  
  if len(s) <= 1:
    return s
  
  return s[-1] + ReverseString(s[:-1])
if __name__ == "__main__":
  test_strings = ["Cambridge", "Computer", "Python", "A", ""]

  for text in test_strings:
    reversed_text = ReverseString(text)
    print(f"Original: '{text}' | Reversed: '{reversed_text}'")
```
