```python
# ==========================================
# MDEN SCHOOL OF PROGRAMMING P4 CHEAT CODE
# Question 1: Doubly Linked List with OOP
# "Talk is cheap, show me the code."
# ==========================================

# (a) Define the Node class
class Node:
    def __init__(self, book_id, title):
        self.BookID = book_id       # Integer
        self.Title = title          # String
        self.Previous = None        # Pointer to preceding Node
        self.Next = None            # Pointer to succeeding Node


# (b) Define the DoublyLinkedList class
class DoublyLinkedList:
    def __init__(self):
        self.StartPointer = None    # Head pointer
        self.TailPointer = None     # Tail pointer

    def InsertNode(self, new_book_id, new_title):
        new_node = Node(new_book_id, new_title)
        
        # Edge Case 1: Inserting into an empty list
        if self.StartPointer is None:
            self.StartPointer = new_node
            self.TailPointer = new_node
            return

        # Edge Case 2: Inserting at the beginning of the list (new ID is smaller than head)
        if new_book_id < self.StartPointer.BookID:
            new_node.Next = self.StartPointer
            self.StartPointer.Previous = new_node
            self.StartPointer = new_node
            return

        # Edge Case 3 & 4: Inserting in the middle or at the end of the list
        current = self.StartPointer
        while current is not None and current.BookID < new_book_id:
            current = current.Next

        # If current is None, we are inserting at the end of the list
        if current is None:
            new_node.Previous = self.TailPointer
            self.TailPointer.Next = new_node
            self.TailPointer = new_node
        else:
            # Inserting in the middle of the list
            new_node.Previous = current.Previous
            new_node.Next = current
            current.Previous.Next = new_node
            current.Previous = new_node

    def OutputNodes(self):
        print("--- Traversing Forward (Start to End) ---")
        current = self.StartPointer
        while current is not None:
            print(f"Book ID: {current.BookID}, Title: {current.Title}")
            current = current.Next

    def OutputReverse(self):
        print("--- Traversing Reverse (End to Start) ---")
        current = self.TailPointer
        while current is not None:
            print(f"Book ID: {current.BookID}, Title: {current.Title}")
            current = current.Previous


# (c) Main Program / Test Harness
if __name__ == "__main__":
    # Instantiate an object of the DoublyLinkedList class
    my_library = DoublyLinkedList()

    # Insert at least 4 books in random order
    my_library.InsertNode(105, "Data Structures in Python")
    my_library.InsertNode(101, "Algorithms Unlocked")
    my_library.InsertNode(108, "Machine Learning Basics")
    my_library.InsertNode(103, "Database Systems")

    # Call OutputNodes() and OutputReverse() to demonstrate correct ordered insertion
    my_library.OutputNodes()
    print()
    my_library.OutputReverse()
```
