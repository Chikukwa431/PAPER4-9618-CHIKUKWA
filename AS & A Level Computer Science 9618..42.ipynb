```python
# BY CHIKUKWA NEWTON
#  Question 2 – Binary Tree using OOP
#  Cambridge International AS & A Level Computer Science 9618/42
#  May/June 2024
# ─────────────────────────────────────────────────────────────


# ─── 2(a)(i)  Class Node – declaration and constructor ───────

class Node:
    # Attribute declarations:
    # LeftPointer  : INTEGER  – index of left child in the tree array (-1 = none)
    # Data         : INTEGER  – the node's data value
    # RightPointer : INTEGER  – index of right child in the tree array (-1 = none)

    def __init__(self, data):
        self.Data = data          # initialise Data to parameter value
        self.LeftPointer = -1     # initialise LeftPointer  to -1
        self.RightPointer = -1    # initialise RightPointer to -1

    # ─── 2(a)(ii)  Get methods ────────────────────────────────

    def GetLeft(self):
        return self.LeftPointer

    def GetRight(self):
        return self.RightPointer

    def GetData(self):
        return self.Data

    # ─── 2(a)(iii)  Set methods ───────────────────────────────

    def SetLeft(self, left):
        self.LeftPointer = left

    def SetRight(self, right):
        self.RightPointer = right

    def SetData(self, data):
        self.Data = data


# ─── 2(b)(i)  Class TreeClass – declaration and constructor ──

class TreeClass:
    # Attribute declarations:
    # Tree        : ARRAY[0:19] OF Node – array of 20 Node objects
    # FirstNode   : INTEGER             – index of the root node (-1 = empty)
    # NumberNodes : INTEGER             – how many nodes have been inserted

    def __init__(self):
        self.FirstNode = -1       # tree is empty
        self.NumberNodes = 0
        # initialise every element to a Node with data value -1
        self.Tree = [Node(-1) for _ in range(20)]

    # ─── 2(b)(ii)  InsertNode() ──────────────────────────────

    def InsertNode(self, NewNode):
        # Store NewNode at the next free position
        self.Tree[self.NumberNodes] = NewNode

        if self.FirstNode == -1:
            # Tree is empty – this node becomes the root
            self.NumberNodes += 1
            self.FirstNode = 0
        else:
            # Tree is not empty – find the correct position
            self.NumberNodes += 1
            Current = self.FirstNode      # start at root
            Placed = False

            while not Placed:
                if NewNode.GetData() < self.Tree[Current].GetData():
                    # Go left
                    if self.Tree[Current].GetLeft() == -1:
                        # Empty left slot – place here
                        self.Tree[Current].SetLeft(self.NumberNodes - 1)
                        Placed = True
                    else:
                        Current = self.Tree[Current].GetLeft()
                else:
                    # Go right
                    if self.Tree[Current].GetRight() == -1:
                        # Empty right slot – place here
                        self.Tree[Current].SetRight(self.NumberNodes - 1)
                        Placed = True
                    else:
                        Current = self.Tree[Current].GetRight()

    # ─── 2(b)(iii)  OutputTree() ─────────────────────────────

    def OutputTree(self):
        if self.NumberNodes == 0:
            print("No nodes")
        else:
            for i in range(self.NumberNodes):
                left  = self.Tree[i].GetLeft()
                data  = self.Tree[i].GetData()
                right = self.Tree[i].GetRight()
                print(f"Index {i}: LeftPointer={left}  Data={data}  RightPointer={right}")


# ─── 2(c)(i)  Main program – declare instance of TreeClass ───

TheTree = TreeClass()

# ─── 2(c)(ii)  Insert integers and call OutputTree() ─────────

values = [10, 11, 5, 1, 20, 7, 15]

for v in values:
    NewNode = Node(v)
    TheTree.InsertNode(NewNode)

TheTree.OutputTree()
```
