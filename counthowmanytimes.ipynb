```python
def CountX(Arr, Low, High, Target):
  if Low > High:
    return 0
  match = 1 if Arr[Low] == Target else 0
  return match + CountX(Arr, Low + 1, High, Target)
def main():
    sample_arr = [1, 2, 3, 2, 4, 2, 5]
    target = 2
    
    result = CountX(sample_arr, 0, len(sample_arr) - 1, target)
    
    print(f"Array: {sample_arr}")
    print(f"Target {target} appears {result} times.")

if __name__ == "__main__":
    main()
```
