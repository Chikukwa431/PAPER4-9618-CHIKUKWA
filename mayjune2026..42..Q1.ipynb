```python
# NEWTON CHIKUKWA WISE OWL MARONDERA
# Question1_J26
# Cambridge International AS & A Level Computer Science 9618/42 Paper 4
# Question 1 - Python (console mode)

# (a) Declare and initialise TheData in the required order
TheData = [
    "laptop",
    "tablet",
    "printer",
    "computer",
    "processor",
    "monitor",
    "memory",
    "cable",
    "touchscreen"
]
# (b)(i) Procedure to find and output longest and shortest strings
def FindLongestShortest(arr):
    # Initialise using the first element
    shortest = arr[0]
    longest = arr[0]

    shortest_len = len(arr[0])
    longest_len = len(arr[0])

    # Scan through all strings and update the extremes using length comparisons
    for s in arr:
        s_len = len(s)

        if s_len < shortest_len:
            shortest = s
            shortest_len = s_len

        if s_len > longest_len:
            longest = s
            longest_len = s_len

    # Output with appropriate messages
    print("Longest string:", longest)
    print("Shortest string:", shortest)
# (c)(i) Function to sort array into ascending alphabetical order using insertion sort
def InsertionSort(arr):
    # Make a copy so original list contents are not destroyed
    a = arr[:]

    for i in range(1, len(a)):
        key = a[i]
        j = i - 1

        # Shift elements of the sorted portion that are greater than key
        while j >= 0 and a[j] > key:
            a[j + 1] = a[j]
            j -= 1

        # Insert key into correct position
        a[j + 1] = key

    return a


# ---- Main program ----
print("Question 1(b): Longest and shortest")
FindLongestShortest(TheData)

print("\nQuestion 1(c): Insertion sort")
print("Before sorting")
print(TheData)

sortedData = InsertionSort(TheData)

print("After sorting")
print(sortedData)
```
