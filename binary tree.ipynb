```python
# Global array declarations
ArrayNodes = [[-1 for i in range(3)] for j in range(20)]
FreeNode = 6
RootPointer = 0

def SetupTree():
    global FreeNode, RootPointer
    # Populate the array with the provided binary tree data[cite: 2]
    ArrayNodes[0] = [1, 20, 5]
    ArrayNodes[1] = [2, 15, -1]
    ArrayNodes[2] = [-1, 3, 3]
    ArrayNodes[3] = [-1, 9, 4]
    ArrayNodes[4] = [-1, 10, -1]
    ArrayNodes[5] = [-1, 58, -1]
    ArrayNodes[6] = [-1, -1, -1]

def SearchValue(Root, ValueToFind):
    if Root == -1:
        return -1
    else:
        if ArrayNodes[Root][1] == ValueToFind:
            return Root
        else:
            if ArrayNodes[Root][1] == -1:
                return -1
            
    if ArrayNodes[Root][1] > ValueToFind:
        return SearchValue(ArrayNodes[Root][0], ValueToFind)
        
    if ArrayNodes[Root][1] < ValueToFind:
        return SearchValue(ArrayNodes[Root][2], ValueToFind)
        
    return -1

def PostOrder(Root):
    if Root != -1:
        # Visit left node
        if ArrayNodes[Root][0] != -1:
            PostOrder(ArrayNodes[Root][0])
        # Visit right node
        if ArrayNodes[Root][2] != -1:
            PostOrder(ArrayNodes[Root][2])
        # Output current root node
        print(ArrayNodes[Root][1])

# --- Main Program ---
SetupTree()
SearchResult = SearchValue(RootPointer, 15)
if SearchResult != -1:
    print(f"Value 15 found at index: {SearchResult}")
else:
    print("Value 15 was not found in the tree.")

print("\nPost-order traversal output:")
PostOrder(RootPointer)
```
