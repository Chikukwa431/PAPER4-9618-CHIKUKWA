```python
# NEWTON CHIKUKWA
# Question 3 - Linked List (Python console mode)

# 2D array LinkedList[index][0]=data, LinkedList[index][1]=pointer
LinkedList = [
    [9512, 5],   # 0
    [5541, 3],   # 1
    [1999, 0],   # 2
    [6651, -1],  # 3
    [-1, 6],     # 4
    [5612, 1],   # 5
    [-1, -1],    # 6
    [-1, 4]      # 7
]

# Pointers with global scope
StartLinkedList = 2
StartEmptyList = 7


def PrintListItems():
    current = StartLinkedList
    while current != -1:
        print(LinkedList[current][0])
        current = LinkedList[current][1]


def PrintEmptyList():
    current = StartEmptyList
    while current != -1:
        print(current)
        current = LinkedList[current][1]


def AddItem(item):
    global StartLinkedList, StartEmptyList

    # If there is no space in the linked list
    if StartEmptyList == -1:
        print("List is full")
        return

    # Insert into the space identified by StartEmptyList
    new_index = StartEmptyList

    # Put item into that node's data
    LinkedList[new_index][0] = item

    # Advance StartEmptyList to next empty node
    StartEmptyList = LinkedList[new_index][1]

    # Insert the new node at the end of the linked list
    if StartLinkedList == -1:
        StartLinkedList = new_index
        LinkedList[new_index][1] = -1
    else:
        current = StartLinkedList
        while LinkedList[current][1] != -1:
            current = LinkedList[current][1]
        LinkedList[current][1] = new_index
        LinkedList[new_index][1] = -1


# ---------- Main program (tests) ----------
print("Linked list")
PrintListItems()

print("Empty list")
PrintEmptyList()

print("Adding item 8451...")
AddItem(8451)

print("New linked list")
PrintListItems()

print("New empty list")
PrintEmptyList()
```
