```python

# def SumOfDigits(n):
# Declares a function named SumOfDigits that accepts a single parameter n, 
# representing the positive integer whose digits will be added together.
# if n < 10:
# Acts as the base case check, determining if n has been reduced to a single-digit number (less than 10) to prevent infinite recursion.
# return n
# Terminates the recursive chain by returning the single-digit value of n directly back up the call stack when the base case condition is met.
# return (n % 10) + SumOfDigits(n // 10)
# Executes the recursive step: n % 10 isolates the rightmost (last) digit, and SumOfDigits(n // 10)
# performs integer division by 10 to remove that last digit, adding the isolated digit to the result of the next recursive call.

def SumOfDigits(n):
  if n < 10:
    return n
  return (n % 10) + SumOfDigits(n // 10)


if __name__ == "__main__":
  test_numbers = [5, 42, 123, 9876, 0]

  for number in test_numbers:
    result = SumOfDigits(number)
    print(f"Number: {number} | Sum of Digits: {result}")
```
