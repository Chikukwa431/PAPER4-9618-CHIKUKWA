```python
# ==================================================
# Cambridge 9618 Paper 4: MDEN School of Programming
# ==================================================

def FindMaximum(DataList, Index):
    """
    Recursively finds the maximum integer value in the array.
    """
    # Base case: if we are at the last element of the list
    if Index == len(DataList) - 1:
        return DataList[Index]
    
    # Recursive step: find maximum of the remaining sub-list
    MaxOfRest = FindMaximum(DataList, Index + 1)
    
    # Compare current element with the max of the rest
    if DataList[Index] > MaxOfRest:
        return DataList[Index]
    else:
        return MaxOfRest


def FindMaximumIterative(DataList):
    """
    Iteratively finds the maximum integer value in the array.
    """
    if not DataList:
        return None
    
    MaxValue = DataList[0]
    for i in range(1, len(DataList)):
        if DataList[i] > MaxValue:
            MaxValue = DataList[i]
            
    return MaxValue


def CountOccurrences(DataList, Index, Target):
    """
    Recursively counts how many times a Target value appears in the array.
    """
    # Base case: if index goes beyond the list bounds
    if Index == len(DataList):
        return 0
    
    # Evaluate current match and add to the result of the rest of the list
    Match = 1 if DataList[Index] == Target else 0
    return Match + CountOccurrences(DataList, Index + 1, Target)


def CalculateSum(DataList, Index):
    """
    Recursively calculates the sum of all elements in the array.
    """
    # Base case: if index goes beyond the list bounds
    if Index == len(DataList):
        return 0
    
    # Recursive step: current element + sum of remaining elements
    return DataList[Index] + CalculateSum(DataList, Index + 1)


# ==========================================
# Main Program Driver
# ==========================================
if __name__ == "__main__":
    # Initialize DataList with the required 10 values[cite: 2]
    DataList = [14, 25, 14, 42, 8, 19, 42, 31, 14, 7]
    
    print("--- Array Analysis Output ---")
    
    # Call FindMaximum starting at index 0 and print result[cite: 2]
    max_val = FindMaximum(DataList, 0)
    print(f"Maximum value (Recursive): {max_val}")
    
    # Test iterative equivalent for verification
    max_val_iter = FindMaximumIterative(DataList)
    print(f"Maximum value (Iterative): {max_val_iter}")
    
    # Call CountOccurrences looking for target 14 starting at index 0[cite: 2]
    target_count = CountOccurrences(DataList, 0, 14)
    print(f"Occurrences of target 14: {target_count}")
    
    # Call CalculateSum starting at index 0 and print result[cite: 2]
    total_sum = CalculateSum(DataList, 0)
    print(f"Sum of all elements: {total_sum}")
```
